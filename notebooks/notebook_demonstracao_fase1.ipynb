## Contrato metodologico

- Target oficial: `finish_position`.
- Metricas oficiais: MAE, RMSE, R2 e Kendall tau.

- As tabelas usadas aqui ja foram geradas pelo pipeline de modelagem.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import Image, display

BASE_DIR = Path.cwd()
REPORTS_DIR = BASE_DIR / 'reports' / 'modelagem'
FIGURES_DIR = REPORTS_DIR / 'figures' / 'semana3'

resumo_path = REPORTS_DIR / 'tabela_metricas_tunadas_4modelos_resumo.csv'
folds_path = REPORTS_DIR / 'tabela_metricas_tunadas_4modelos.csv'

resumo = pd.read_csv(resumo_path, encoding='utf-8-sig')
folds = pd.read_csv(folds_path, encoding='utf-8-sig')
```

## Tabela final dos modelos

```python
cols = [
    'modelo', 'mae_medio', 'mae_std', 'rmse_medio', 'rmse_std',
    'r2_medio', 'kendall_tau_medio', 'score_composto_medio',
    'tempo_tuning_minutos'
]
resumo[cols].sort_values('score_composto_medio', ascending=False)
```

Leitura esperada: Ridge e o melhor modelo/baseline global; LightGBM e a melhor arvore; Random Forest fica muito proxima; XGBoost permanece como comparativo relevante.

```python
melhor_global = resumo.sort_values('score_composto_medio', ascending=False).iloc[0]
arvores = resumo[resumo['modelo'].isin(['lightgbm_tuned', 'random_forest_tuned', 'xgboost_tuned'])]
melhor_arvore = arvores.sort_values('score_composto_medio', ascending=False).iloc[0]

print(f"Melhor global: {melhor_global['modelo']} | score={melhor_global['score_composto_medio']:.4f} | MAE={melhor_global['mae_medio']:.4f}")
print(f"Melhor arvore: {melhor_arvore['modelo']} | score={melhor_arvore['score_composto_medio']:.4f} | MAE={melhor_arvore['mae_medio']:.4f}")
```
